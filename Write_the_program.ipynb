```python
n = int(input("enter a number"))
original = n
rev = 0
while n > 0:
    digit = n % 10
    rev = rev *10 + digit
    n=n//10
if original == rev:
    print("Palindrome")
else:
    print("Not Palindrome")

```

```python
#include <stdio.h>

int main() {
    for (int i = 2; i <= 30; i++) {
        int prime = 1;

        for (int j = 2; j < i; j++) {
            if (i % j == 0) {
                prime = 0;
                break;
            }
        }

        if (prime == 1) {
            printf("%d ", i);
        }
    }

    return 0;
}
```
